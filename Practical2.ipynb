Handling Missing Values

(a)Mean Imputation (StudyHours)

```python
print(df['StudyHours'].mean())
df['StudyHours'] = df['StudyHours'].fillna(df['StudyHours'].mean())
```

(b) Median Imputation (Marks)

```python
print(df['Marks'].median())
df['Marks'] = df['Marks'].fillna(df['Marks'].median())
```

(c) Mode Imputation (if needed – categorical)

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

(d) IQR Method (Outlier Detection – Marks)

```python
Q1 = df['Marks'].quantile(0.25)
Q3 = df['Marks'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Cap outliers
df['Marks'] = np.where(df['Marks'] < lower, lower,
                        np.where(df['Marks'] > upper, upper, df['Marks']))
df['Marks']
```

e)Standard deviation

```python
df['Marks'].std()
```

Label Encoding (Gender)

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

print(df[['Gender']].head())
```

Dummy Variables (Department)

```python
df = pd.get_dummies(df, columns=['Department'], drop_first=True)

print(df.head())
```

Normalization (Min-Max Scaling)

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Age', 'StudyHours', 'Marks']] = scaler.fit_transform(
    df[['Age', 'StudyHours', 'Marks']]
)

print(df.head())
```

Scaling (Standardization)

```python
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()

df[['Age', 'StudyHours', 'Marks']] = std_scaler.fit_transform(
    df[['Age', 'StudyHours', 'Marks']]
)

print(df.head())
```
